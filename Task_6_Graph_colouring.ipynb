```python
import matplotlib.pyplot as plt
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D'],
    'C': ['A', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}
def get_chromatic_number(graph):
    nodes, edges = list(graph.keys()), [(u, v) for u in graph for v in graph[u] if u < v]
    def solve(k, assignment={}):
        if len(assignment) == len(nodes):
            return assignment
        node = next(n for n in nodes if n not in assignment)
        for c in range(k):
            if all(assignment.get(nbr) != c for nbr in graph[node]):
                res = solve(k, {**assignment, node: c})
                if res: return res
        return None
    for k in range(1, len(nodes) + 1):
        coloring = solve(k)
        if coloring:
            return k, coloring
min_colors, coloring = get_chromatic_number(graph)
print(f"Minimum colors: {min_colors}")
print(f"Color assignment: {coloring}")
plt.show()
```

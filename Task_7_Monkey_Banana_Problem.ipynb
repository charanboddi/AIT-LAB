```python

def monkey_banana():
    monkey = "door"
    box = "window"
    bananas = "ceiling"
    monkey_on_box = False
    has_banana = False
    print("Initial State:")
    print(f"Monkey: {monkey}, Box: {box}, Bananas: {bananas}")
    print("\n1. Monkey walks to the box.")
    monkey = box
    print("2. Monkey pushes the box under the bananas.")
    box = bananas
    print("3. Monkey climbs onto the box.")
    monkey_on_box = True
    if monkey_on_box and monkey == box:
        print("4. Monkey grabs the bananas.")
        has_banana = True
    print("\nFinal State:")
    if has_banana:
        print("Monkey has the bananas! ")
    else:
        print("Monkey could not get the bananas.")
monkey_banana()
```
